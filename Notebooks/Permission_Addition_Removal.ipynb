**Add User to Access Package**

```python
import requests

tenant_id = '705787a3-c501-46b9-9d55-84e1ea8b7e90'
client_id = '277274ae-b302-44a6-b2d6-24d707daa8ce'
client_secret = ' '

user_id = "8fa85648-4c9a-486d-9d90-39f59f254806"  #john@MngEnvMCAP866405.onmicrosoft.com
access_package_id = "61478292-23f6-4fc5-8bf5-7b063fbba7db" #Cements_OneSales_Team
assignment_policy_id = "9dce05b6-4b62-492a-93cb-9c9caed5c3ac" # Initial Policy

# Step 1: Get Token
token_url = f"https://login.microsoftonline.com/{tenant_id}/oauth2/v2.0/token"

token_data = {
    "grant_type": "client_credentials",
    "client_id": client_id,
    "client_secret": client_secret,
    "scope": "https://graph.microsoft.com/.default"
}

token_response = requests.post(token_url, data=token_data)
access_token = token_response.json()["access_token"]

# Step 2: Call Graph API
url = "https://graph.microsoft.com/v1.0/identityGovernance/entitlementManagement/assignmentRequests"

headers = {
    "Authorization": f"Bearer {access_token}",
    "Content-Type": "application/json"
}

payload = {
    "requestType": "AdminAdd",
    "assignment": {
        "targetId": user_id,
        "accessPackageId": access_package_id,
        "assignmentPolicyId": assignment_policy_id
    }
}

response = requests.post(url, json=payload, headers=headers)

print("Status:", response.status_code)
print("Response:", response.text)
```

**Get assignmentId and remove the assignment**

```python
assignments_url = f"https://graph.microsoft.com/v1.0/identityGovernance/entitlementManagement/assignments?$expand=accessPackage,target&$filter=accessPackage/id eq '{access_package_id}' and target/objectid eq '{user_id}' and status eq 'Delivered'"
#print(assignments_url)
response = requests.get(assignments_url, headers=headers)
data = response.json()
#print(data)

assignment_id = data["value"][0]["id"] if data.get("value") else None
print("Assignment ID:", assignment_id)

if assignment_id:
    remove_payload = {
        "requestType": "AdminRemove",
        "assignment": {
            "id": assignment_id
        }
    }

    response = requests.post(url, json=remove_payload, headers=headers)

    print("Remove Status:", response.status_code)
    print("Response:", response.text)
else:
    print("No assignment found for removal")
```
