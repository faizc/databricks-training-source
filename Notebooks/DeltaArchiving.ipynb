```scala
%python
import pyarrow.parquet as pq
from datetime import datetime
from dateutil import parser

date1 = datetime(2020, 4, 1)
date2 = datetime(2020, 5, 31)

filesReadyForMovement = []
filesWithConflict = []

all_files = dbutils.fs.ls("/mnt/delta-table/yellow-taxi/")
files = []
for fileinfo in all_files:
    if(fileinfo.name.endswith('.parquet')):
        #print(fileinfo.name)
        files.append(fileinfo.name)
#print(date)
#print(files)
# test with single file 
#files.append('part-00129-5f432c31-2234-4ba8-96bd-20f1e2eb4cbd-c000.snappy.parquet')
#files.append('part-00000-b0add483-3f3d-4760-b959-27e6ba055c8f-c000.snappy.parquet')
for fileName in files:
    filePath='/dbfs/mnt/delta-table/yellow-taxi/'+fileName
#    print(filePath)
    parquet_file = pq.ParquetFile(filePath)
    pfile = pq.read_table(filePath)

    #print(pfile.schema)
    colName = 'tpep_pickup_datetime'
    #print("Field Index : {}".format(pfile.schema.get_field_index(colName)))
    colIndex = pfile.schema.get_field_index('tpep_pickup_datetime')
    numRowGroups = parquet_file.metadata.num_row_groups
    #print('colName = ', colName, ' ColIndex = ', colIndex)
    #print('numRowGroups = ', numRowGroups)
    fileWithinRange = False
    cond = -1

    for rowGroup in range(0, numRowGroups):
        #print('rowGroup =', rowGroup, '= b')
        #print(parquet_file.metadata.row_group(rowGroup).column(colIndex))
        min = parser.parse(parquet_file.metadata.row_group(rowGroup).column(colIndex).statistics.min)
        max = parser.parse(parquet_file.metadata.row_group(rowGroup).column(colIndex).statistics.max)
        #print('min =', type(min), ' max = ', type(max))
        minFlag = min>date1
        maxFlag = max<date2
        if(minFlag and maxFlag):
            fileWithinRange = True
            cond = 1
        else:
            fileWithinRange = False
            cond = 2
            break
    
    if(fileWithinRange):
        filesReadyForMovement.append(fileName)
    else:
        filesWithConflict.append(fileName)
    
    print('min =', parquet_file.metadata.row_group(rowGroup).column(colIndex).statistics.min,
          ' max = ', parquet_file.metadata.row_group(rowGroup).column(colIndex).statistics.max,
         'minFlag = ', minFlag,
         'maxFlag = ', maxFlag,
         'fileName = ', fileName,
         'cond = ', cond)
    print('\n')

#    dbutils.fs.mv("source-file", "targetfolder")

# Less than year is hot and in between (1-3) range is cold and >3 is archive 

# Define different combinations to move the data 
# 1. Find data between the data range and move to desired folder 
# 2. Find data less than the max and move to archive folder (Archive)
# 3. Find data that is between ranges and segregate the relevant data and then move to desired folder 

# Be Careful 
# 1. For delta table, directly moving the data would corrupt the state. It would be better to VACCUM the table so that all the changes are committed and there are no versions. Setting the VACCUM to zero can cause some unknown behavior https://docs.microsoft.com/en-us/azure/databricks/kb/delta/data-missing-vacuum-parallel-write
# 2. On safer side, perform deep cloning so that table state is not lost 

# Data Lifecycle Management 

# Separate Process 
# 1. Deep Copy of the table, VACCUM and then we can delete the delta-logs folder (non-delta table)
# 2. Archive the parquet files as per the script (500) - (100 - Archive) (100 - Hot) (300 - Cold)  
# 3. Read the hot parquet files and create the delta table based on the data validation
```
