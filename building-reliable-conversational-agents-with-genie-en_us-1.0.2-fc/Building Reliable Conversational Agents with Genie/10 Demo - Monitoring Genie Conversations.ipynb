# Demo - Monitoring Genie Conversations

## Overview

In the previous demo, you stepped into the role of a business user by asking questions and rating responses. Now it is time to switch back to the **curator perspective** and review that feedback.

The **Monitoring** tab gives you visibility into every question asked, the SQL Genie generated, and the feedback users provided. This is where you identify gaps and prioritize your next round of curation.

## Learning Objectives

By the end of this demo, you will be able to:

1. **Navigate the Monitoring tab** to review user questions, ratings, and generated SQL.
2. **Filter monitoring data** by rating type (thumbs up, thumbs down, request review).
3. **Apply the 4-step curator workflow** (Triage, Diagnose, Fix, Verify) to process monitoring feedback.
4. **Follow monitoring best practices** for ongoing Genie Space improvement.
5. **Understand the continuous improvement loop** that connects user feedback to curation actions.

| **Rating** | Thumbs up, thumbs down, or request review (if rated) |

   a. Select the **Requested for review question**. 
      - Should be the question **Which franchise locations have the worst customer reviews?**. This is the question that was rated as a **Request Review** in an earlier demonstration.

   b. Scroll through the question and user request.

   c. Notice as the curator you can:
      - **Add a comment for the user**

      - **Mark as reviewed**

   d. Add a comment and select **Mark as reviewed**.

   e. Navigate back to **Monitoring** and remove the filter on the questions. 
   
   f. Notice now that the question is marked as **Reviewed** under the **Request** column.

### A4. Monitoring Best Practices

<div style="
  border-left: 4px solid #1976d2;
  background: #e3f2fd;
  padding: 14px 18px;
  border-radius: 4px;
  margin: 16px 0;
">
  <strong style="display:block; color:#0d47a1; margin-bottom:6px; font-size: 1.1em;">
    Best Practices for Monitoring
  </strong>
  <div style="color:#333;">

- **Check Monitoring regularly** - Set a cadence (daily at first, then weekly) to review new questions and feedback. Do not wait for complaints.
- **Prioritize thumbs-down and request-review** - These are your users telling you where Genie needs help. Start here every time.
- **Look for patterns, not one-offs** - If multiple users ask similar questions that Genie gets wrong, that is a systemic gap worth fixing. A single unusual question may not need a Knowledge Store change.
- **Turn monitoring insights into benchmarks** - When you find a question that Genie answered incorrectly, add it as a benchmark with ground truth SQL. This prevents the same failure from recurring after future changes.
- **Review thumbs-up responses too** - Occasionally check positive-rated responses to confirm they are actually correct. Users sometimes rate quickly without verifying the SQL.
- **Track improvement over time** - As you add more curation, the ratio of thumbs-up to thumbs-down should improve. Monitoring is how you measure that trend.
