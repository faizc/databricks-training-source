## D. Add a SQL Expression to the Knowledge Store

SQL expressions let you define **derived columns**. 

That is, values that do not exist in the data but can be computed from existing columns. 

Genie can use these expressions in generated SQL.

### D1. Add a Region SQL Expression to the Knowledge Store

Our business groups franchise countries into three specific regions:
  - **APJ**
  - **EMEA**
  - **AMER**

This mapping does not exist as a column, we need to teach it to Genie as a SQL expression.

<div style="
  border-left: 4px solid #ff9800;
  background: #fff3e0;
  padding: 14px 18px;
  border-radius: 4px;
  margin: 16px 0;
">
  <strong style="display:block; color:#e65100; margin-bottom:6px; font-size: 1.1em;">
    Benchmark This Fixes
  </strong>
  <div style="color:#333;">

**Benchmark #5: What are total sales by region?** - Genie most likely groups by `country` because there is no `region` column. This expression creates the mapping.

  </div>
</div>

1. In your Genie Space, select **Configure** > **Instructions** > **SQL Expressions**.

2. Select **+ Add** > **Field**.

3. **Name the Field:** `region`

4. Run the cell below to generate your specific **SQL Expression** (adds your unique catalog name in the expression for you).
    - **NOTES:** 
        - When referencing the column in the expression, we are using a three level-namespace + the column name
        - `YOUR_CATALOG.genie_course_bakehouse.sales_franchises_gold.country`

````python
%sql
DECLARE OR REPLACE my_query STRING DEFAULT "";

SET VAR my_query = "
CASE
    WHEN " || my_catalog || ".genie_course_bakehouse.sales_franchises_gold.country IN ('Japan', 'Australia') 
      THEN 'APJ'
    WHEN " || my_catalog || ".genie_course_bakehouse.sales_franchises_gold.country IN ('US', 'Canada') 
      THEN 'AMER'
    WHEN " || my_catalog || ".genie_course_bakehouse.sales_franchises_gold.country IN ('Netherlands', 'France', 'Germany', 'Italy', 'Sweden')
      THEN 'EMEA' 
    ELSE 'Other'
END
";

SELECT my_query;
````

5. Copy the output above and paste into the **Code** field and review the query.

6. Add the following:
    - **Synonyms**: `franchise region, region sales`
    - **Instructions**: `When user asks for metrics by 'region' where region is defined by business mapping of country to APJ, AMER, EMEA, or Other; SCOPE_TABLES: Use with sales_franchises_gold and sales_transactions_gold, grouping or reporting at the franchise-location level; RISK_IF_MISUSED: Do not use unless the user explicitly refers to regions per this mapping (e.g., could mislead if a different regional breakdown is implied). This is a company-specific mapping.`

7. Select **Save**.

### E1. Add Example Query: Who are our best customers?

Teach Genie what query to use when you get the best customers question.

**Benchmark #3: Who are our best customers?** — Genie currently guesses what "best" means, adds extra columns, and does not limit results. This instruction locks in the definition.

  </div>
</div>

1. In your Genie Space, select **Configure** > **Instructions** > **SQL Queries**.

2. Select **+ Add** > **Example Query**.

3. Enter the question: `Who are our best customers?`

4. Run the cell below to generate the query with your catalog name, then paste it as the SQL query.

````python
%sql
DECLARE OR REPLACE my_query STRING DEFAULT "";

SET VAR my_query = "

SELECT
  c.customerID,
  c.first_name,
  c.last_name,
  SUM(t.totalPrice) AS total_spend
FROM
  " || my_catalog || ".genie_course_bakehouse.sales_transactions_gold AS t
    JOIN " || my_catalog || ".genie_course_bakehouse.sales_customers_gold AS c
      ON t.customerID = c.customerID
GROUP BY
  c.customerID,
  c.first_name,
  c.last_name
ORDER BY
  total_spend DESC
LIMIT 10
";

SELECT my_query;
````

### E2. Add Example Query: Franchises per Supplier (LEFT JOIN Pattern)

This teaches Genie the `LEFT JOIN + COALESCE` pattern needed to include franchises that do not have a matching supplier.

<div style="
  border-left: 4px solid #ff9800;
  background: #fff3e0;
  padding: 14px 18px;
  border-radius: 4px;
  margin: 16px 0;
">
  <strong style="display:block; color:#e65100; margin-bottom:6px; font-size: 1.1em;">

  </strong>
  <div style="color:#333;">

**Benchmark #4: How many franchises does each supplier serve?** - Genie currently uses an INNER JOIN, silently dropping 21 of 48 franchises. This example teaches the correct LEFT JOIN pattern.

  </div>
</div>

1. In your Genie Space, select **Configure** > **Instructions** > **SQL Queries**.

2. Select **+ Add**.

3. Enter the question: `How many franchises does each supplier serve?`

4. Run the cell below to generate the query with your catalog name, then paste it as the SQL answer.

````python
%sql
DECLARE OR REPLACE my_query STRING DEFAULT "";

SET VAR my_query = "
SELECT
    COALESCE(s.name, 'No Matching Supplier') AS supplier_name,
    COUNT(f.franchiseID) AS franchise_count
FROM " || my_catalog || ".genie_course_bakehouse.sales_franchises_gold AS f
LEFT JOIN " || my_catalog || ".genie_course_bakehouse.sales_suppliers_gold AS s
    ON f.supplierID = s.supplierID
GROUP BY
    s.name
ORDER BY
    franchise_count DESC
";

SELECT my_query;
````

5. Copy the output above and paste into the **SQL query** field.

6. Select **Preview** and review the results. They should look like this:

function copyBlock() {
  const el = document.getElementById("copy-block");
  if (!el) return;

  const text = el.innerText;

  // Preferred modern API
  if (navigator.clipboard && navigator.clipboard.writeText) {

function fallbackCopy(text) {
  const textarea = document.createElement("textarea");
  textarea.value = text;
  textarea.style.position = "fixed";
  textarea.style.left = "-9999px";
  document.body.appendChild(textarea);
  textarea.select();

}
