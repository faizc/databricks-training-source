# Demo - Sharing and the End User Experience with Genie

## Overview

You have built and curated a Genie Space through descriptions, SQL logic, instructions, and benchmarks. Before end users can interact with your Genie, you need to understand **permissions**, **share** the space, and see what the experience looks like from the **end-user perspective**.

In this demo, you will first review permissions and share your Genie Space. Then you will switch to the **end-user role** by opening **Genie** (formerly Databricks One), asking natural language questions, and rating responses to simulate real-world feedback. Finally, you will see how **Unity Catalog governance** automatically protects data even when accessed through your Genie Space.

## Learning Objectives

By the end of this demo, you will be able to:

1. **Understand the required permissions** to create or edit a Genie Space (entitlements, compute, data, ACLs).
2. **Describe the four Genie ACL levels** (CAN VIEW, CAN RUN, CAN EDIT, CAN MANAGE) and their capabilities.
3. **Share a Genie Space** with users or groups and select the appropriate permission level.
4. **Access a Genie Space through Genie** as an end user would.
5. **Ask natural language questions** and rate responses using thumbs up, thumbs down, and request review.
6. **Observe that Genie respects Unity Catalog governance**, including ABAC column masks, without any Genie-specific configuration.

## E. Unity Catalog Governance with Genie

You have shared your Genie Space and seen how end users interact with it. 

But how do you ensure that **sensitive data stays protected** when users ask questions?

#### Genie Spaces does not bypass Unity Catalog permissions.

- Any **ABAC policies**, **row filters**, **column masks**, or **table ACLs** applied to the underlying tables are automatically enforced when Genie generates and executes SQL.

- This means you can safely share a Genie Space with business users, knowing that they will only see data they are authorized to access.

### E1. Applying an ABAC Policy to a Genie Source Table

Your organization wants to centrally manage PII protection across all tables without applying individual masks one by one.

We'll set this up in three SQL steps, then ask Genie about <code>email_address</code> to prove the policy is enforced automatically.

    <div style="font-size: 15pt; font-weight: 700; color: #0b2026;">Attach an ABAC policy</div>
    <div style="font-size: 14pt; color: #5A6F77; margin-top: 4px;">Policy auto-masks any tagged column across the schema</div>
  </div>

1. Run the cell below to create a masking function and apply it to the `email_address` column.
    - This function masks email addresses for everyone except members of the `email_admins` group. 
    - Since you are not in that group, **you will see masked emails**.

```sql
-- 1. Apply the system governed tag to identify this column for ABAC policies
ALTER TABLE sales_customers_gold
ALTER COLUMN email_address SET TAGS ('class.email_address');

-- 2. Create a masking UDF for ABAC policy
CREATE OR REPLACE FUNCTION genie_course_bakehouse.abac_mask_email(email_value STRING)
RETURNS STRING
RETURN CONCAT('***@', SPLIT(email_value, '@')[1]);

-- 3. The POLICY's TO clause determines which groups are affected and binds everything together
CREATE OR REPLACE POLICY mask_email_policy
  ON SCHEMA genie_course_bakehouse
  COMMENT 'Mask email addresses for general users across all tagged tables'
  COLUMN MASK abac_mask_email TO `account users`
  FOR TABLES MATCH COLUMNS hasTag('class.email_address') AS email_col ON COLUMN email_col;
```
