# Demo - Creating and Exploring a Genie Space

## Overview

In this demo, you will create a Genie Space using the **bakehouse** tables. This is a fictional bakery franchise data you explored in the previous demo. 

You will configure the Genie Space, explore the UI, and interact with Genie using a series of targeted natural language questions. The goal is to observe how Genie behaves out of the box, before any metadata, instructions, or example SQL have been added. 

This baseline experience will motivate the improvements covered in later demos.

## Learning Objectives

By the end of this demo, you will be able to:

1. **Create a Genie Space** by selecting Unity Catalog tables and configuring a SQL warehouse.
2. **Navigate the Genie Space UI** including the chat interface, settings panel, and data tab.
3. **Interpret Genie-generated SQL** to evaluate whether responses are correct and business-aligned.
4. **Identify limitations of an unconfigured Genie Space** such as ambiguous business terms, unknown column values, and complex join patterns.
5. **Provide feedback on Genie responses** using thumbs up/down ratings and request review to flag answers for improvement.

<div class="step s1 questions">
  <div class="step-header">
    <div class="step-number">1</div>
    <div class="step-title">Start with purpose and questions</div>
  </div>
  <div class="step-text">
    <ul>
      <li>Define a <strong>narrow topic</strong> and <strong>clear audience</strong>
        <ul>
          <li><i>Example: "Sales pipeline for NA sales managers"</i></li>
        </ul>
      </li>
      <li>Write <strong>5–10 real business questions</strong> this audience asks</li>
      <li>These will become your <strong>sample questions</strong> and <strong>benchmarks</strong></li>
    </ul>
    <div class="example">
      <strong>Bakehouse example:</strong>
      <ul>
        <li><b>Audience:</b> Bakehouse franchise managers</li>
        <li>How Many Customers Do We Have?</li>
        <li>What are our top 5 franchise locations by revenue?</li>
        <li>Who are our best customers?</li>
        <li>Count of Bad Reviews by Location?</li>
        <li>How Many Franchises Does Each Supplier Serve?</li>
        <li>What Are Total Sales by Region?</li>
      </ul>
    </div>
  </div>
</div>

<div class="step s2 tables">
  <div class="step-header">
    <div class="step-number">2</div>
    <div class="step-title">Select a minimal data slice</div>
  </div>
  <div class="step-text">
    <ul>
      <li>Work backwards from your questions to identify the <strong>smallest set of tables or views</strong> needed</li>
      <li>Aim for <strong>3–5 tables or views</strong></li>
      <li>Stay well under the <strong>30-table limit</strong></li>
      <li>Prefer <strong>pre-joined or metric views</strong> (metric views are outside the scope of this course) to:
        <ul>
          <li>Hide join complexity</li>
          <li>Encode business logic once</li>
        </ul>
      </li>
    </ul>
    <div class="example">
      <strong>Bakehouse example:</strong>
      <ul>
        <li><code>sales_customers_gold</code> - customer profiles and locations</li>
        <li><code>sales_franchises_gold</code> - franchise locations and size</li>
        <li><code>sales_suppliers_gold</code> - ingredient suppliers</li>
        <li><code>sales_transactions_gold</code> - sales transactions and revenue</li>
        <li><code>media_customer_reviews_gold</code> - reviews and sentiment</li>
      </ul>
    </div>
  </div>
</div>

<div class="step s3 metadata">
  <div class="step-header">
    <div class="step-number">3</div>
    <div class="step-title">Add metadata and synonyms</div>
  </div>
  <div class="step-text">
    <ul>
      <li>Ensure names and descriptions reflect <strong>business language</strong></li>
      <li><strong>Hide</strong> confusing or rarely used columns</li>
      <li>Add <strong>synonyms</strong> so Genie understands how users speak
        <ul>
          <li>Example: "rep" → "salesperson"</li>
        </ul>
      </li>
    </ul>
    <div class="example">
      <strong>Bakehouse example:</strong>
      <ul>
        <li><code>sentiment</code> - add valid values: <code>positive</code>, <code>negative</code>, <code>mixed</code></li>
        <li><code>country</code> - disambiguate customer country vs. franchise country</li>
        <li>Synonym: "revenue" > <code>totalPrice</code></li>
        <li>Synonym: "rating" > <code>sentiment</code></li>
      </ul>
    </div>
  </div>
</div>

<div class="step s4 instructions">
  <div class="step-header">
    <div class="step-number">4</div>
    <div class="step-title">Encode business logic in SQL</div>
  </div>
  <div class="step-text">
    <ul>
      <li>Create <strong>SQL expressions</strong> for core metrics and filters</li>
      <li>Add high-quality <strong>example SQL queries</strong></li>
      <li><strong>Parameterize</strong> inputs like region, product, and dates</li>
    </ul>
    <div class="example">
      <strong>Bakehouse example:</strong>
      <ul>
        <li>Example SQL: "Top products by revenue" using <code>SUM(totalPrice)</code></li>
        <li>Example SQL: <code>LEFT JOIN</code> + <code>IS NULL</code> to find franchises without a valid supplier</li>
        <li>SQL expression: revenue = <code>SUM(totalPrice)</code></li>
      </ul>
    </div>
    <div class="callout">
      <strong>Guidance:</strong> SQL expressions and example queries are the primary way Genie learns your business logic.
    </div>
  </div>
</div>

<div class="step s5 global">
  <div class="step-header">
    <div class="step-number">5</div>
    <div class="step-title">Add light global text instructions (last resort)</div>
  </div>
  <div class="step-text">
    <ul>
      <li>Use for <strong>clarification behavior</strong> and <strong>formatting rules</strong></li>
      <li>Keep instructions <strong>short and specific</strong></li>
      <li>Ensure <strong>consistency</strong> with SQL logic and examples</li>
      <li>Text instructions are only for context that should be applied <strong>globally</strong> and do not fit into the other formats</li>
      <li>Think of text instructions as <strong>last resorts</strong> for any general guidances</li>
    </ul>
    <div class="example">
      <strong>Bakehouse example:</strong>
      <ul>
        <li>"Best customer" = highest total spend using <code>SUM(totalPrice)</code></li>
        <li>Always use <code>ILIKE</code> for text matching</li>
        <li>Round all currency values to 2 decimal places</li>
        <li>Default to franchise country when asked about location</li>
      </ul>
    </div>
    <div class="callout">
      Avoid duplication or contradictions with your SQL definitions.
    </div>
  </div>
</div>

<div class="step s6 testing">
  <div class="step-header">
    <div class="step-number">6</div>
    <div class="step-title">Test, benchmark, and iterate</div>
  </div>
  <div class="step-text">
    <ul>
      <li>Ask your <strong>original questions</strong> and inspect <strong>generated SQL</strong></li>
      <li>Turn strong questions into <strong>benchmarks</strong></li>
      <li>Run a <strong>pilot</strong> with real users and gather feedback</li>
      <li>Use <strong>Monitoring</strong> to identify failure points</li>
      <li><strong>Continuously refine</strong> data, metadata, SQL, and instructions</li>
    </ul>
    <div class="example">
      <strong>Bakehouse example:</strong>
      <ul>
        <li>Re-ask all 5 original questions after each improvement</li>
        <li>Add benchmark: "Top 5 franchises by revenue" with expected SQL using <code>SUM(totalPrice)</code></li>
        <li>Check the Monitoring tab for thumbs-down responses</li>
        <li>Iterate on instructions and metadata based on failures</li>
      </ul>
    </div>
    <div class="callout">
      This is where great Genie spaces are actually built, through iteration.
    </div>
  </div>
</div>

  <div class="summary-title">Key Takeaway</div>
  <div class="summary-text">
    Start narrow, test early, and iterate often. <br></br>The best Genie spaces are not built in one pass, they are refined through real usage and feedback.

   [GCP](https://docs.databricks.com/gcp/en/genie/benchmarks)

   > **NOTE:** Benchmark questions run as new conversations. They do not carry the same context as a threaded Genie conversation. Each question is processed as a new query, using the instructions defined in the space, including any provided example SQL and SQL functions.

   c. **Monitor**
   - See how the space is used.
   - View all questions, ratings, feedback and problem cases.
   - Use it to decide what to fix or improve.

   - Shows how Genie has been "taught."

In this first demo, we are **not** adding any of the following to the Genie Space:

- **Table or column metadata** - descriptions and synonyms on the tables that help Genie understand what each table and column represents

- **Sample questions** - example queries that help Genie understand common question patterns

- **Text instructions** - behavioral guidance like formatting rules or clarification prompts
