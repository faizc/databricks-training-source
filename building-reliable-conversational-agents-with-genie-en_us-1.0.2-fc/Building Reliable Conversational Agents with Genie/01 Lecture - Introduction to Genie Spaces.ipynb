# Lecture - Introduction to Genie Spaces

## Overview

A **Genie Space** is a curated, natural language interface to a domain's Unity Catalog data, enabling business users to interact with that data using plain English. It uses generative AI tailored to your organization's data and terminology, and it can be continuously improved through monitoring and user feedback.

Rather than requiring business users to ask data analysts for help, Genie Spaces allow anyone to ask questions in plain English and receive SQL-backed answers, tables, and visualizations, all grounded in governed Unity Catalog data.

In this lecture, you will quickly review the components of the Databricks AI/BI product family:

- AI/BI Dashboards
- Genie Spaces
- Genie *(formerly Databricks One)*
- Genie Code

You will learn how they complement each other, then focus on how a Genie Space translates a natural language question into a SQL result, understand the components that make up a Genie Space, and consider the common challenges that arise when building one without additional configuration.

## Learning Objectives

By the end of this lecture, you will be able to:

1. **Describe the Databricks AI/BI product family** and explain the difference between them.
2. **Explain how a Genie Space translates a natural language question into a SQL result**, including the five-step process from question to answer.
3. **Identify and describe the six components of a Genie Space**: Unity Catalog tables, SQL logic (joins, expressions, example queries), text instructions, common questions, table/column metadata, and benchmarks.
4. **Recognize common challenges in Genie Space configuration** like misunderstood business jargon, incorrect table or column usage, and opaque column values.

##### EXPAND FOR ADDITIONAL NOTES

<details>

Databricks AI/BI is a product family that brings intelligence to business analytics. It includes four complementary tools:

#### AI/BI Dashboards
Dashboards are the traditional BI experience.

Analysts build curated visualizations, charts, and reports that stakeholders consume. Dashboards are great when you know exactly what questions need to be answered and want to present that data in a polished, repeatable format.

#### Genie Spaces
Genie Spaces take the opposite approach. Instead of pre-building answers, a Genie Space lets business users **ask their own questions** in plain English. The Space interprets the question, generates SQL, runs it against your Unity Catalog tables, and returns the result.

All without the user needing to know SQL.

#### Genie
Genie is a user interface designed for business users, giving them a single, intuitive entry point to interact with data and AI in Databricks, without needing to navigate technical concepts such as compute, queries, models, or notebooks. **Genie was previously called Databricks One.**

#### Genie Code
Genie Code is the AI assistant for builders. It lives inside notebooks, Lakeflow pipelines, the SQL editor, and the dashboard editor, helping data engineers, analysts, and data scientists write, debug, and iterate on code.

#### How They Work Together
- **Dashboards** answer the questions you **know** your business will ask repeatedly.
- **Genie Spaces** answer the questions that come up **on the fly**. Ad-hoc questions that no one anticipated when designing the dashboard.
- **Genie** fills in the gaps between tools and people by bringing your data, dashboards, and AI assistants into one place, so everyone can find, understand, and act on insights in their daily work.


Together, they give organizations both curated insights and self-service exploration on the same governed data.

</details>

<details>

This flow shows the core idea behind Genie Spaces.

A user asks a question in plain, everyday language. No SQL, no dashboards to navigate, no need to know where the data lives.

A Genie Space takes that question, understands the intent, and uses your governed data to generate a response.

That response can include queries, summaries, and visualizations.

### Consumers
- Can ask questions the same way you would ask a colleague
- A Genie Space translates your question into data queries behind the scenes
- Results are grounded in your organization's data, not generic answers
- You can iterate, refine, and explore without needing technical skills

Think of it as moving from "I need to write SQL" to "I just ask what I want to know."

</details>

<details>

High-level Genie Space overview.

#### 1. User Asks a Question
The user types a question in plain English into the Genie Space chat interface. No SQL knowledge is required.

Questions can range from:

- Simple: "How many customers do we have?"
- Complex: "What are the top 5 franchise locations by revenue in Japan?"



#### 2. Genie Selects Relevant Context
Genie does not just "read everything". It selects the most relevant context to understand the question.

This can include:

- **Table schemas** — column names and data types from Unity Catalog tables
- **Column and table descriptions** — metadata explaining what fields represent
- **SQL logic** — joins, expressions, and example queries that encode business definitions
- **Instructions** — guidance for behavior, edge cases, and business rules
- **Example questions** — natural language to SQL patterns
- **Conversation history** — prior questions in the same thread

The better this context is curated, the more accurate the Genie Space becomes.



#### 3. SQL Is Generated
Based on the selected context, Genie generates a **read-only SQL query**.

- The SQL reflects both the user question and the curated data model
- It is fully visible to the user, there is no black box
- In some cases, Genie may reuse trusted or predefined SQL logic

Users can review the SQL to validate correctness.



#### 4. SQL Is Executed
The generated SQL runs on the Genie space's configured SQL warehouse.

- Handles scale and concurrency automatically
- Uses the same execution engine as Databricks SQL
- Returns structured results from governed data



#### 5. Response Is Returned
- Results are displayed as a table, often with a short summary
- The generated SQL is shown for transparency
- Users can provide feedback (thumbs up/down)
- Follow-up questions can be asked in the same conversation

If Genie cannot confidently answer, it may first ask a **clarifying question** before generating SQL.

</details>

##### EXPAND FOR ADDITIONAL NOTES

<details>

A Genie Space is made up of six context components. Only tables are required.

#### Unity Catalog Tables (Required)
You select Unity Catalog tables and views to give Genie its default data context, but Genie is not strictly limited to only those objects. It can query other tables when users prompt for joins, edit SQL directly, or when outside tables are referenced in instructions or metadata. Access is governed by Unity Catalog permissions, so the real boundary is data entitlement, not just the list of tables attached to the space.


#### Table and Column Metadata (Strongly Recommended)
Descriptions and synonyms at table/column level, stored in Unity Catalog and/or the Genie **Knowledge Store**.

- Critical for mapping natural language terms to the right columns and values
- Disambiguates similar fields (e.g., customer `country` vs. franchise `country`)
- For example, a column named `size` with values `S`, `M`, `L`, `XL` is ambiguous without a description
- Adding "Franchise size category. Values: S (Small), M (Medium), L (Large), XL (Extra Large)" tells Genie exactly how to interpret it
- The Knowledge Store also provides **prompt matching** capabilities (format assistance and entity matching) that help Genie map user language to actual data values

> **Note:** The Knowledge Store is covered in detail in the next lecture. For now, think of it as the centralized place within a Genie Space where you configure all curation settings.


#### SQL Logic: Joins, Expressions, Queries (Highly Recommended)
This is the primary way Genie learns your business logic. Docs explicitly say to **prioritize SQL expressions and example SQL over text instructions** for business semantics.

- **Joins / relationships** — Prefer PK/FK constraints in UC or pre-joined views/metric views rather than describing joins in text
- **SQL expressions / functions** — Define business terms and metric calculations in SQL (e.g., revenue = `SUM(totalPrice)`), or register them as UC functions
- **Example SQL queries** — Pair natural language questions with the SQL that answers them. Genie uses these as trusted patterns when generating queries

#### Instructions (Text) - Behavior & Edge Cases (Recommended, but "last resort")
Use text instructions for things that **cannot be expressed cleanly in SQL or metadata**:

- When to ask clarifying questions
- How to format answers (rounding, language, row limits)
- What to do when users omit key filters (time range, region, etc.)

**Avoid** using Instructions to define metrics, joins, or filters if you can express them in SQL instead.


#### Common Questions (Recommended)
Example business questions that teach Genie common question patterns. These can serve two purposes:

1. **Guide users** — Common questions appear in the UI as suggestions for users, showing users what kinds of questions they can ask  
2. **Guide Genie** — You can take common questions and add them as **SQL Queries** to act as examples that help Genie generate better SQL

**Best practice:** Attach SQL answers when possible so they become example queries, not just free-floating questions.


#### Benchmarks (Strongly Recommended)
Structured test suites that let you measure and track Genie's accuracy over time.

- Save key questions with optional **ground truth SQL** that defines the expected correct answer
- Run benchmarks after every change (descriptions, instructions, example SQL) to confirm improvements and catch regressions
- Each benchmark question runs as a **new, independent conversation** — no thread context is carried over, simulating a fresh user interaction
- Essential for the **test, benchmark, and iterate** cycle that turns a good Genie Space into a great one

> **Note:** Benchmarks do **not** change how Genie answers questions. They are purely for evaluation and quality tracking.

</details>

## E. Conclusion

### E1. Key Concepts Summary

In this lecture, you explored the Databricks AI/BI product family and the role of Genie Spaces in enabling self-service data exploration. Key concepts covered include:

1. **AI/BI components** - Dashboards deliver curated, repeatable answers to known questions. Genie Spaces enable business users to ask ad-hoc questions in natural language without writing SQL.
2. **Genie** - Genie is a user interface designed for business users, giving them a single, intuitive entry point to interact with data and AI in Databricks, without needing to navigate technical concepts such as compute, queries, models, or notebooks.
3. **The five-step Genie Space process** - A user asks a question, the Genie Space interprets it using available context, generates SQL, executes it on a configured SQL warehouse, and returns a transparent, data-backed result.
4. **The main components of a Genie Space** - Unity Catalog tables (required), SQL logic including joins, expressions, and example queries (highly recommended), text instructions for behavior and edge cases, common questions, table/column metadata, and benchmarks for structured testing and iteration. Optional components significantly improve answer quality.
5. **Common configuration challenges** - Ambiguous business terms, shared column names across tables, and opaque column values can all cause a Genie Space to misinterpret questions without additional guidance.
