# Lecture - Curating an Effective Genie Space

## Overview

In the previous demo, you created a Genie Space with no metadata, instructions, or example SQL and observed how Genie can struggle with ambiguous business terms, shared column names, and opaque column values. Those gaps are not limitations of Genie itself, they are gaps in how the space was configured.

In this lecture, you will learn how to think like a **space curator**, someone who iteratively improves a Genie Space by adding the right metadata and guidance. 

You will explore the Knowledge Store as the centralized curation surface, understand the recommended order for applying curation tools, and connect each tool back to the specific problems you observed in the previous demo.

## Learning Objectives

By the end of this lecture, you will be able to:

1. **Describe the iterative curation process** for building an effective Genie Space: build, test, observe, improve, repeat.
2. **Recall the six curation components** from the previous lecture and connect them to Knowledge Store configuration.
3. **Explain the Knowledge Store** as the centralized configuration surface for curation, including its components: space-level metadata, prompt matching, join relationships, SQL expressions, and knowledge mining.
4. **Explain the recommended order of operations** for adding metadata to a Genie Space and why descriptions come first.
5. **Map each problem observed in the previous demo** to the specific curation tool that addresses it.
6. **Apply table selection best practices** to ensure the foundation of a Genie Space supports accurate and consistent query generation.

A Genie Space is not something you build once and walk away from.

#### It is a product that you design, test, and refine over time, just like a dashboard or a report.

##### EXPAND FOR ADDITIONAL NOTES

<details>

Building an effective Genie Space requires two distinct roles:

#### The Space Curator
The curator is the person who builds and maintains the Genie Space. This person needs two things:
1. **Domain knowledge** - They understand what the data means, how metrics are calculated, and what terminology the business uses.
2. **SQL proficiency** - They can write the example queries and instructions that teach Genie how to interpret questions correctly.

Data analysts are often the ideal curators because they sit at the intersection of business knowledge and technical skill.

#### The Space Consumer
- The consumer is the end user, the person asking questions.

- They should not need to know SQL, understand table schemas, or worry about which column to filter on.

- Their experience is only as good as the curation behind the space.

#### Summary
The quality of a Genie Space is determined by the **curator**, not the consumer.

If consumers are getting wrong answers, the fix is better curation, not better questions.

</details>

##### EXPAND FOR ADDITIONAL NOTES

<details>

The curation loop is the most important concept in this lecture. Here is how it works:

#### 1. Create
Start with the minimum viable space. Add only the tables needed to answer your target set of questions. 

Do not try to add every instruction or description upfront. You will not know what is needed until you test.

#### 2. Test
Ask questions that you expect your business users to ask. 

Start with simple questions and work toward more complex ones. This is exactly what we did in the previous demo.

#### 3. Observe
After each question, **review the generated SQL carefully**. Do not just look at the result. Look at *how* Genie arrived at the result. Ask yourself:
- Did Genie use the right table?
- Did it join correctly?
- Did it filter on the right column values?
- Did it interpret the business term the way I intended?

#### 4. Improve
Based on what you observed, add the appropriate metadata. Then go back to Step 2 and re-ask the same questions **in a new chat thread** to confirm the improvement.

#### Why a New Chat Thread?
Genie loads instructions when a new chat starts. After updating instructions, always test in a **new chat** to ensure the latest configuration is being used.

#### Start Small
It is impossible to define a fully complete space on the first try. Start with a small set of target questions and a minimal set of tables. A focused space with 5 well-described tables will outperform a broad space with 20 poorly documented ones.

</details>

##### EXPAND FOR ADDITIONAL NOTES

<details>

The Knowledge Store is where you configure how Genie understands your data and responds to questions.

It brings together multiple layers of context that guide how Genie interprets natural language and generates SQL.



#### 1. Space-Level Metadata
This is where you define **descriptions and synonyms** for tables and columns.

- Helps map business language to the correct data fields
- Clarifies ambiguous columns (for example, customer country vs franchise country)
- Can extend or override Unity Catalog metadata without modifying the underlying data

This is one of the **highest-impact areas for improving accuracy**.


#### 2. Prompt Matching
Prompt matching improves how Genie connects user input to actual data values.

- **Format assistance** provides representative values for columns
- **Entity matching** maps user terms to real values in the data

This is especially important for **categorical filters** like region, product names, or status fields.



#### 3. SQL Joins
Defines how tables relate to each other.

- Specify join conditions explicitly
- Set relationship types (one-to-many, many-to-one, etc.)

Without clear joins, Genie may generate incorrect or incomplete queries.



#### 4. SQL Expressions
This is where you define reusable business logic.

- Measures and KPIs (for example, revenue = SUM(totalPrice))
- Filters and reusable conditions
- Dimensions used for grouping

This is the **preferred way to encode business definitions**, rather than using text instructions.



#### 5. SQL Queries (Example Queries and Functions)
These teach Genie how to answer real business questions.

- Pair a **natural language question** with the SQL that answers it
- Can be static or **parameterized** for reuse
- Genie can reuse these directly or learn patterns from them
- SQL functions can be registered in Unity Catalog for reusable, trusted logic

This is one of the most effective ways to improve **accuracy and consistency**.



#### 6. General Instructions
Text-based guidance that applies globally across all queries.

- Capture **business context** and domain-specific rules
- Control formatting and response behavior (language, rounding, output style)
- Apply rules that cannot be expressed in SQL or metadata

**Important:** Instructions should be used as a **last resort**, not for defining metrics or joins.



### Key takeaway

Genie performs best when you prioritize:

1. **Metadata** for meaning  
2. **SQL logic and queries** for business definitions  
3. **Instructions only when necessary**

Think of the Knowledge Store as a system where:

- Structured inputs (SQL, metadata) drive accuracy  
- Text instructions provide fallback guidance  

</details>

##### EXPAND FOR ADDITIONAL NOTES

<details>

The order matters. Each layer builds on the previous one.


#### 1. Define Purpose and Scope
Start by deciding what the Genie Space is for.

- Define the audience and business topic
- Identify the key questions the space should answer
- Keep the scope narrow and focused

A Genie Space should support a specific use case, not every possible question across the business.



#### 2. Add Focused Unity Catalog Tables and Relationships
Once the purpose is clear, select the data that supports it.

- Use a small, focused set of Unity Catalog tables or views
- Ensure **relationships (joins)** are defined (PK/FK, views, or Knowledge Store)
- Prefer well-documented, analytics-ready tables
- Aim for about **5 tables or fewer** when possible
- Pre-join or simplify data with views or metric views if needed

Without clean data and clear relationships, the rest of the curation work is much harder.



#### 3. Add Metadata and Prompt Matching
Metadata is the **strongest early accuracy lever** and often the easiest win.

- Add **descriptions and synonyms** to clarify what columns mean
- Resolve ambiguity across similar fields
- Help Genie select the right columns when questions are vague

Also enable **prompt matching**:

- **Format assistance** helps Genie understand data formats and value patterns
- **Entity matching** maps user language, including misspellings, to real data values

Many spaces improve significantly at this stage before any SQL is added.



#### 4. Add SQL Expressions
SQL expressions define your reusable business logic.

- Define metrics and KPIs
- Create reusable filters and dimensions
- Standardize how business terms are calculated

Recommended to use SQL expressions for **common, reusable definitions** instead of text.



#### 5. Add Example SQL Queries
Example SQL queries teach Genie how to answer common or complex questions.

- Pair natural language questions with verified SQL
- Use them for multi-step logic or ambiguous prompts
- Parameterized examples can be reused as patterns

Genie can use these examples directly or learn from them to answer similar questions.



#### 6. Add General Instructions
Use text instructions only when something **cannot be expressed in SQL or metadata**.

Good use cases include:

- Clarification rules
- Formatting and output behavior
- Language or response preferences

**Avoid using instructions to define metrics, joins, or logic.**

---

#### A Practical Guideline

- **Start small** and expand over time
- Use **clear column names and descriptions**
- Focus on a few high-value questions first
- Prioritize **structured context** like metadata and SQL over plain text guidance

</details>

##### EXPAND FOR ADDITIONAL NOTES

<details>

#### Why Table Quality Matters
No amount of instructions can compensate for poorly structured input data. If your column names are cryptic, your data is inconsistent, or your tables require complex transformations to be useful, Genie will struggle regardless of how much guidance you provide.

#### Best Practices
- **Use gold tables** - curated, analytics-ready tables that represent the final layer of your data pipeline. This is what we did in the previous demo with the `_gold` suffix tables.
- **Keep tables focused** - A space with 5 well-chosen tables will outperform a space with 20 loosely related ones. The more tables you add, the more likely Genie is to pick the wrong one.
- **Define relationships** - If foreign key relationships are defined in Unity Catalog, Genie can use them to determine how tables join together. If they are not defined, document the relationships in instructions.
- **Clear column names** - If a column is named `size`, that is ambiguous. A column named `franchise_size_category` is self-documenting. When you cannot rename columns, add descriptions instead.
- **Don't auto-generate descriptions** - AI-generated column comments can be generic or misleading. Only add descriptions that provide context Genie would not otherwise have.

#### When to Use Views
If your data model is complex with many self-joins, complicated aggregations, or columns that need transformation consider creating a **view** that simplifies the data for Genie. 

**A pre-joined, pre-aggregated view can dramatically reduce the complexity of the SQL Genie needs to generate.**

</details>

## Where to Invest Time for the Biggest Impact

When building a Genie space, not all effort produces equal results.  

Some improvements significantly increase accuracy and usability, while others add complexity without much benefit.

Focus your time on the areas that have the **largest impact on how Genie understands and answers questions**.


### High-Impact Levers

These are the areas where your effort will most improve Genie performance:

- **Start with clear business questions and a focused scope** - Define the key questions you want Genie to answer. Avoid trying to support everything at once.

- **Well-described tables and columns using business language** - Add clear, human-readable descriptions. Genie relies heavily on this context to interpret user questions correctly.

- **Example SQL aligned to real usage patterns** - Provide sample queries that reflect how users actually ask questions. These act as patterns Genie can learn from.

- **Simple, consistent joins with a clear row-level meaning** - Make relationships between tables easy to follow, and ensure each table clearly represents one type of record. Avoid joins that unintentionally duplicate or distort results.


### Low-Impact (Common Traps)

These are common areas where effort often does not translate into better outcomes:

- **Overly long or complex instructions** - More text does not mean better results. Keep instructions concise and focused.

- **Adding too many tables upfront** - A large, unfocused dataset makes it harder for Genie to choose the right data.

- **Over-optimizing edge cases early** - Focus on the most common and important questions first. Handle edge cases later.

- **Treating Genie as 'set and forget'**  - Genie improves through iteration. Regularly review results and refine your setup.

---

### Key Takeaway

Genie works best when you treat it like a product:

- Start small  
- Focus on clarity  
- Improve based on real usage  

Investing in the right areas early will lead to **more accurate, consistent, and trustworthy results**.

## E. Conclusion

In this lecture, you learned how to approach Genie Space curation as an iterative, structured process:

- A Genie Space is a **product** that requires ongoing testing and refinement by a domain-expert curator.
- The **iterative curation loop**: create, test, observe, improve
  - This is the core workflow for building an effective space.
