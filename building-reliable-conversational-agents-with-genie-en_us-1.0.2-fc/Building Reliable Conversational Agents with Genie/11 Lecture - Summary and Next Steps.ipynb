##### EXPAND FOR ADDITIONAL NOTES

<details>

#### The Complete Lifecycle

This visual shows the full authoring lifecycle you completed across all demos. Each stage builds on the previous one.

#### Unity Catalog Tables

Everything starts with governed data. You selected 5 gold tables from the bakehouse dataset. In production, work backwards from your stakeholder questions to identify the smallest set of tables needed.

#### Knowledge Store

This is where the majority of curation work happens. You added four types of context:

- **Metadata, Synonyms, and Display Names** give Genie foundational understanding of what each table and column means. They disambiguate similar columns, connect user language to column names, and provide clear display labels.
- **Prompt Matching** teaches Genie the actual values in your data so it filters correctly in WHERE clauses.
- **SQL Logic (Expressions, Joins, Queries)** encodes business logic that Genie can't infer from column names. Derived dimensions like region, complex join patterns, and metric definitions all live here.
- **General Instructions** handle the edge cases and business rules that don't fit in SQL or metadata.

#### Genie Space

The configured space is the product of everything you put into the Knowledge Store. It's a curated conversational analytics experience, not just a chatbot on top of tables.

#### Genie (formerly Databricks One)

This is where your work reaches its audience. End users interact with your Genie Space through Genie without ever seeing the Knowledge Store or the curation work behind it. They ask questions in plain language, get SQL-backed answers, and provide feedback through ratings.

#### Monitoring

Monitoring closes the feedback loop. You see what users actually ask (not just what you planned for), which responses they rated positively or negatively, and where Genie still needs help.

#### Benchmarks

Benchmarks are your test suite. They sit outside the Knowledge Store because they don't teach Genie anything. They measure whether your curation changes actually improved accuracy and catch regressions when you make further updates.

#### The Iterate Loop

The coral bar represents the continuous improvement cycle. Insights from monitoring and benchmark results feed back into Knowledge Store updates. In production, this loop runs indefinitely. Users ask new questions, you review monitoring data, update the Knowledge Store, re-run benchmarks, and the space gets better over time.

</details>
